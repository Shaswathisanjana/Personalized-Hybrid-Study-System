````python
import requests
import xml.etree.ElementTree as ET
from urllib.parse import quote_plus

# -----------------------------
# Semantic Scholar Search
# -----------------------------

SEMANTIC_URL = "https://api.semanticscholar.org/graph/v1/paper/search"

FIELDS = ",".join([
    "title",
    "abstract",
    "authors",
    "year",
    "citationCount",
    "url",
    "venue"
])

def semantic_search(query, limit=10):

    params = {
        "query": query,
        "limit": limit,
        "fields": FIELDS
    }

    try:
        response = requests.get(SEMANTIC_URL, params=params, timeout=20)

        if response.status_code == 429:
            print("⚠️ Semantic Scholar rate limit exceeded. Skipping...")
            return []

        response.raise_for_status()

        data = response.json()

        papers = []

        for paper in data.get("data", []):

            papers.append({
                "title": paper.get("title"),
                "abstract": paper.get("abstract"),
                "authors": [a["name"] for a in paper.get("authors", [])],
                "year": paper.get("year"),
                "citations": paper.get("citationCount"),
                "venue": paper.get("venue"),
                "url": paper.get("url")
            })

        return papers

    except requests.exceptions.RequestException as e:
        print(f"Semantic Scholar Error: {e}")
        return []


# -----------------------------
# arXiv Search
# -----------------------------

def arxiv_search(query, limit=10):

    query = quote_plus(query)

    url = (
        f"https://export.arxiv.org/api/query?"
        f"search_query=all:{query}"
        f"&start=0"
        f"&max_results={limit}"
    )

    try:
        response = requests.get(url, timeout=20)
        response.raise_for_status()

        root = ET.fromstring(response.content)

        ns = {
            "atom": "http://www.w3.org/2005/Atom"
        }

        papers = []

        for entry in root.findall("atom:entry", ns):

            papers.append({
                "title": entry.find("atom:title", ns).text.strip(),
                "summary": entry.find("atom:summary", ns).text.strip(),
                "year": entry.find("atom:published", ns).text[:4],
                "link": entry.find("atom:id", ns).text
            })

        return papers

    except Exception as e:
        print(f"arXiv Error: {e}")
        return []


# -----------------------------
# Main
# -----------------------------

query = input("Research Topic: ")

papers = []

papers.extend(semantic_search(query))
papers.extend(arxiv_search(query))

print(f"\nFound {len(papers)} papers\n")

for i, p in enumerate(papers, 1):

    print("=" * 80)
    print(f"Paper {i}")
    print("Title :", p.get("title"))
    print("Year  :", p.get("year"))
    print("URL   :", p.get("url") or p.get("link"))
````

````python
import requests
import os

def download_pdf(url, folder="papers"):

    os.makedirs(folder, exist_ok=True)

    if "abs" in url:
        url = url.replace("/abs/", "/pdf/") + ".pdf"

    filename = url.split("/")[-1]

    path = os.path.join(folder, filename)

    response = requests.get(url)

    with open(path, "wb") as f:
        f.write(response.content)

    return path
import fitz

def read_pdf(path):

    doc = fitz.open(path)

    text = ""

    for page in doc:

        text += page.get_text()

    return text

import json
client = genai.Client(api_key=GEMINI_API_KEY)



def analyze_paper(text):

    prompt = f"""
You are an expert AI research assistant.

Analyze this research paper.

Return ONLY valid JSON.

If information is unavailable write:
"Not Mentioned"

Schema:

{{
"title":"",
"research_problem":"",
"research_objective":"",
"domain":"",
"dataset":[],
"dataset_size":"",
"preprocessing":[],
"data_augmentation":[],
"model":"",
"backbone":"",
"architecture":"",
"loss_function":"",
"optimizer":"",
"learning_rate":"",
"batch_size":"",
"epochs":"",
"evaluation_metrics":[],
"results":{{
    "accuracy":"",
    "precision":"",
    "recall":"",
    "f1_score":"",
    "map":"",
    "iou":""
}},
"hardware":"",
"advantages":[],
"limitations":[],
"future_work":[],
"research_gap":[],
"novel_contributions":[],
"keywords":[],
"summary":""
}}

Paper:

{text[:50000]}
"""

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite-preview",
        contents=prompt
    )

    output = response.text

    # Remove markdown if Gemini adds it
    output = output.replace("```json", "")
    output = output.replace("```", "")
    output = output.strip()

    try:
        return json.loads(output)

    except:
        return output
            
links = [

    "https://arxiv.org/abs/2109.05746",

    "https://arxiv.org/abs/1902.06197",

    "https://arxiv.org/abs/2008.12589"

]

analyses = []

for link in links:

    pdf = download_pdf(link)

    text = read_pdf(pdf)

    result = analyze_paper(text)

    analyses.append(result)

    print(json.dumps(result, indent=4))

def find_research_gap(all_analysis):

    prompt = f"""
You are a research scientist.

Below are analyses of multiple papers.

Identify:

1. Common approaches

2. Common datasets

3. Common limitations

4. Missing research

5. Research opportunities

6. What can be improved?

7. Which techniques are underexplored?

8. Novel project ideas.

Paper Analyses:

{all_analysis}
"""

    response = client.models.generate_content(
        model="gemini-3.1-flash-lite-preview",
        contents=prompt
    )

    return response.text

gap = find_research_gap(
    json.dumps(analyses, indent=2)
)
print(gap)
````

````python
import pandas as pd


def create_paper_table(analyses):

    rows = []

    for paper in analyses:

        rows.append({

            "Title": paper.get("title"),

            "Year": paper.get("year","Not Mentioned"),

            "Domain": paper.get("domain"),

            "Dataset": ", ".join(paper.get("dataset",[])),

            "Dataset Size": paper.get("dataset_size"),

            "Learning Type": (
                "Unsupervised" 
                if "unsupervised" in str(paper.get("model")).lower()
                else "Supervised"
            ),

            "Model": paper.get("model"),

            "Backbone": paper.get("backbone"),

            "Architecture": paper.get("architecture"),

            "Loss Function": paper.get("loss_function"),

            "Optimizer": paper.get("optimizer"),

            "Epochs": paper.get("epochs"),

            "Batch Size": paper.get("batch_size"),

            "Metrics": ", ".join(
                paper.get("evaluation_metrics",[])
            ),

            "Accuracy": paper.get("results",{}).get("accuracy"),

            "Precision": paper.get("results",{}).get("precision"),

            "Recall": paper.get("results",{}).get("recall"),

            "F1 Score": paper.get("results",{}).get("f1_score"),

            "mAP": paper.get("results",{}).get("map"),

            "Advantages": "; ".join(
                paper.get("advantages",[])
            ),

            "Limitations": "; ".join(
                paper.get("limitations",[])
            ),

            "Research Gap": paper.get("research_gap"),

            "Novel Contribution": "; ".join(
                paper.get("novel_contributions",[])
            )

        })


    return pd.DataFrame(rows)
paper_table = create_paper_table(analyses)

paper_table.to_excel(
    "PCB_Paper_Comparison.xlsx",
    index=False
)

print("Excel saved successfully")
````
